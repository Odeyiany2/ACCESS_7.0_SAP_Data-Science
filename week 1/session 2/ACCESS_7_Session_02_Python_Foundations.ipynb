## 2. Variables: storing information

A **variable** is a name we give to a value so that we can use it later.

Think of a variable as a labelled container. Instead of repeatedly typing a value, we store it once and refer to it by its name.

We use the assignment operator `=` to assign a value to a variable.

````python
student_name = "Ada"
student_age = 21
test_score = 86

print(student_name)
print(student_age)
print(test_score)
````

### Variables can be reused

If a value changes, we can update the variable. Any later calculation that uses it will use the new value.

````python
price = 2500
quantity = 4

total_cost = price * quantity

print(total_cost)
````

Let's update the quantity:

````python
quantity = 6
total_cost = price * quantity

print(total_cost)
````

## 3. Understanding data types

A data type tells Python what kind of value it is working with.

Here are four common types we'll use frequently:

| Data type | Meaning | Example |
|---|---|---|
| `str` | Text (string) | `"Lagos"` |
| `int` | Whole number (integer) | `25` |
| `float` | Number with a decimal | `19.5` |
| `bool` | True or False | `True` |

Let's create examples.

````python
city = 'Lagos'             # string
number_of_students = 35   # integer
average_score = 72.5      # float
assignment_submitted = False  # boolean

print(city)
print(number_of_students)
print(average_score)
print(assignment_submitted)
````

### Text and numbers are not the same

The value `25` is an integer, while `"25"` is a string because it is inside quotation marks.

This distinction matters when performing calculations.

````python
25, 30, 90
"25", "90"
````

````python
age_as_number = 25
age_as_text = "25"

print(type(age_as_number))
print(type(age_as_text))
````

### Converting between data types

Sometimes data comes in as text even when we want to use it as a number. We can convert compatible values.

Common conversion functions include:
- `int()` – converts a compatible value to an integer.
- `float()` – converts a compatible value to a decimal number.
- `str()` – converts a value to text.

````python
quantity = "12"
print(type(quantity))

quantity = float(quantity)

print(quantity)
print(type(quantity))
````

````python
quantity_text = "12"
quantity_number = int(quantity_text)

print(quantity_number)
print(type(quantity_number))
````

````python
price_text = "1500.50"
price_number = float(price_text)

print(price_number)
print(type(price_number))
````

````python
#convert a number to text
number = 20
print(type(number))

number = str(number)
print(number)
print(type(number))
````

**Be careful:** Not every string can be converted into a number. For example, `int("hello")` will produce an error because `"hello"` is not a number.

````python
text = "Hi"
text_2 = int(text)
print(text_2)
````

## 4. Operators: working with values

Operators are symbols or keywords that tell Python to perform an operation.

### A. Arithmetic operators

These are used for calculations.

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `10 + 5` |
| `-` | Subtraction | `10 - 5` |
| `*` | Multiplication | `10 * 5` |
| `/` | Division | `10 / 5` |
| `//` | Floor division | `11 // 2` |
| `%` | Modulus (remainder) | `11 % 2` |
| `**` | Exponentiation | `2 ** 3` |

Let's try them.

````python
a = 10
b = 3


print(type(a))
print(type(b))
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Classic Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

### Example: calculating sales revenue

A shop sells 18 notebooks at ₦1,200 each. Revenue is the selling price multiplied by the quantity sold.

````python
selling_price = 1200
quantity_sold = 18

revenue = selling_price * quantity_sold

print("Total Revenue for Notebooks is :", revenue)
````

### B. Comparison operators

Comparison operators compare two values. The result is a Boolean: either `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

````python
student_score = 68

print(student_score >= 50)
print(student_score == 100)
print(student_score < 40)
print(student_score != 68)
print(student_score == 68)
````

### C. Logical operators

Logical operators allow us to combine conditions.

- `and`: both conditions must be True.
- `or`: at least one condition must be True.
- `not`: reverses a Boolean value.

````python
age = 22
has_student_id = True

print(age >= 18 and has_student_id)
print(age < 18 or has_student_id)
print(not has_student_id)
````

## 5. Conditional statements: teaching Python to make decisions

In everyday life, we make decisions based on conditions.

For example:
- If it rains, take an umbrella.
- If a student's score is at least 50, they pass.
- If stock is below a certain level, reorder the product.

Python uses conditional statements to make decisions.

### The `if` statement

The code inside an `if` block runs only when its condition is True.

````python
temperature = 32

if temperature > 30:
    print("It is a hot day.")
````

````python
temperature = 28

if temperature > 30:
    print("It is a hot day")
````

Notice the colon (`:`) after the condition and the indentation before `print()`. Indentation tells Python which instructions belong inside the condition.

### The `if` and `else` statements

Use `else` when you want Python to choose between two paths.

````python
score = 45

if score >= 50:
    print("You passed.")
else:
    print("You did not pass. Keep practising.")
````

````python
score = 45

if score >= 50:
   print("You passed)
else:
   print("You did not pass.")
````

### The `if`, `elif` and `else` statements

Use `elif` when there are several possible conditions.

````python
score = 76

if score >= 70:
    print("Grade: A")
elif score >= 60:
    print("Grade: B")
elif score >= 50:
    print("Grade: C")
else:
    print("Grade: F")
````

Try changing `score` to `85`, `65`, `52` and `35`. Notice how the output changes.

### Example: checking a simple budget

A student has a budget of ₦15,000 for a purchase.

````python
budget = 15000
purchase_cost = 12500

if purchase_cost < budget:
    print("The purchase is within budget.")
elif purchase_cost == budget:
    print("The purchase uses the entire budget.")
else:
    print("The purchase is above budget.")
````

### Example: checking stock levels

A shop wants to know whether it needs to restock a product.

````python
stock_available = 8
reorder_level = 10

if stock_available <= reorder_level:
    print("Restock this product.")
else:
    print("Stock level is sufficient.")
````

## 6. Loops: repeating instructions

Sometimes we need to perform the same action several times. Instead of writing the same instruction repeatedly, we can use a loop.

Python has two common loop types:
- `for` loops
- `while` loops

Today, we'll have a first look at `for`. But We'll practise them more deeply in our next session.

### A. The `for` loop

A `for` loop repeats an instruction for each item in a sequence.

````python
for number in [1, 2, 3, 4, 5]:
    print(number)
````

````python
stock_list = [1, 2, 3, 4, 5, 6]

for number in stock_list:
    print(number * 2)
````

````python
1 *2
2*2
3*2
````

### Using `range()`

`range()` generates a sequence of numbers. For example, `range(5)` produces the numbers 0, 1, 2, 3 and 4.

````python
for number in range(5):
    print(number)
````

We can also specify a starting and ending point. The ending value is not included.

````python
for number in range(1,10):
    print(number)
````

````python
print(range(1,6))
````

### Example: adding up daily sales

Imagine a shop recorded sales of ₦5,000, ₦7,500, ₦4,000 and ₦6,500 over four days.

````python
daily_sales = [5000, 7500, 4000, 6500]
total_sales = 0

for x in daily_sales:
    total_sales = total_sales + x

print(total_sales)
````

````python
daily_sales = [5000, 7500, 4000, 6500]

total_sales = 0

for sale in daily_sales:
    total_sales = total_sales + sale

print("Total sales: ₦", total_sales)
````

## 7. Functions: reusable blocks of code

A function is a named, reusable block of code that performs a particular task.

Functions help us avoid repeating the same code and make programmes easier to organise.

Python already has built-in functions such as `print()`, `type()` and `len()`.

We can also create our own functions using `def`.

````python
def greet():
    print("Welcome to Data Science!")

greet()
````

### Functions with parameters

A parameter is a name listed in the function definition. It allows us to provide information to the function.

````python
def greet_student(name):
    print("Hello,", name)

greet_student("Ada")
greet_student("Tunde")
````

## Functions that return a value
A function can send a result back using return. This means we can store the result or use it in another calculation

````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

result = calculate_total(1500, 4)
print("Total cost: ₦", result)
````
