## 1. Quick warm-up: loops

A loop allows us to repeat a block of code without writing the same instruction several times.

We will revisit both `for` and `while` loops. Pay particular attention to the `while` loop because it is useful when we want repetition to continue while a condition remains true.

### 1.2 A `while` loop

````python
count = 1

while count <= 5:
    print("Count:", count)
    count = count + 1
````

**Understand the three important parts:**

1. `count = 1` gives the loop a starting value.
2. `while count <= 5` is the condition. Python keeps repeating the block while this condition is true.
3. `count = count + 1` updates the value so the loop can eventually stop.

**Important:** If the condition never becomes false, a `while` loop can run forever. Always check that something inside the loop changes the condition.

| `for` loop | `while` loop |
|---|---|
| Often used to go through items in a collection | Often used when repetition depends on a condition |
| Useful when the number of items or repetitions is known | Useful when we do not know exactly how many repetitions will be needed |
| Moves through items automatically | We usually update the condition ourselves |

### 1.4 A practical `while` loop example

````python
stock = 3

while stock > 0:
    print("One item sold. Stock remaining:", stock)
    stock -= 1

print("Out of stock!")
````

### 1.5 Using `break` and `continue`

````python
# break stops the loop completely
for number in range(1, 8):
    if number == 5:
        break
    print(number)
````

````python
# continue skips the current round and moves to the next one
for number in range(1, 8):
    if number == 5:
        continue #skip 5 and continue
    print(number)
````

### 2.6 Looping through a list

This is where we connect lists with the `for` loops we learned previously.

Instead of writing a separate print statement for every product, we can loop through the list.

````python
products = ["Notebook", "Pen", "Calculator", "School bag"]

for product in products:
    print(product)
````

Let's understand what happens:

1. Python takes the first item in the list.
2. It temporarily stores that item in the variable `product`.
3. It executes the indented instruction.
4. It moves to the next item.
5. It repeats until every item has been processed.

The name `product` is not special. We could use `item`, `name` or another meaningful variable name.

````python
# Combining a list, a loop and a conditional

prices = [1200, 350, 5500, 8000]

for x in prices:
  if x >= 5000:
    print("Price is above 5000:", x)
  else:
    print("Price is below 5000:", x)
````

## 3. Dictionaries: organising related information

A list is useful when we want to store several items.

But what if we want to store different details about one item?

For example, a product may have:

- A name
- A price
- A quantity sold
- Remaining stock

A **dictionary** stores information in key-value pairs.

- A **key** is the label used to identify a piece of information.
- A **value** is the information stored under that key.

Dictionaries are created using curly brackets `{}`.

````python
product = {
    "name": "Notebook",
    "price": 1200,
    "quantity_sold": 18,
    "stock": 8
}

print(product)
# print(type(product))
````

````python
print(product["stock"])
````

````python
print(product["inventory"])
````

### 3.1 Accessing dictionary values

We use the key to retrieve its corresponding value.

Unlike lists, we do not need to remember the numerical position of each piece of information.

````python
print(product["name"])
print(product["price"])
print(product["quantity_sold"])
````

If we request a key that does not exist using square brackets, Python raises a `KeyError`.

We can also use `.get()`. This allows us to provide a default value when a key is missing.

````python
print(product.get("name"))
print(product.get("category"))

print(product.get("category", "Category not provided"))
````

````python
print(product.get("Inventory", 0))
````

### 3.2 Updating and adding information

Dictionaries are mutable. We can update existing values and add new key-value pairs.

````python
product = {
    "name": "Notebook",
    "price": 1200,
    "quantity_sold": 18,
    "stock": 8
}

# Update an existing value
product["price"] = 1300

# Add a new key-value pair
product["Inventory"] = 500
product["category"] = "Stationery"

print(product)
````

### 3.3 Removing information

We can use `pop()` to remove a key and its associated value.

````python
product = {
    "name": "Notebook",
    "price": 1200,
    "quantity_sold": 18,
    "stock": 8
}

removed_stock = product.pop("price")

print("Removed stock:", removed_stock)
print(product)
````

### 3.4 Looping through a dictionary

There are three useful dictionary methods:

| Method | What it returns |
|---|---|
| `.keys()` | All the keys |
| `.values()` | All the values |
| `.items()` | Both keys and values |

````python
product = {
    "name": "Notebook",
    "price": 1200,
    "quantity_sold": 18,
    "stock": 8
}

for k in product.keys():
  print(k)

for v in product.values():
  print(v)


for k, v in product.items():
  print(k, ":", v)

# print("Keys:")
# for key in product.keys():
#     print(key)

# print("Values:")
# for value in product.values():
#     print(value)

# print("Keys and values:")
# for key, value in product.items():
#     print(key, ":", value)
````

## 4. Functions: writing reusable instructions

Imagine we need to calculate the revenue of 100 different products.

We could write the same multiplication repeatedly, but that would make our code unnecessarily long.

Instead, we can create a **function**.

A function is a reusable block of code designed to perform a particular task.

We write the instructions once and call the function whenever we need it.

### 4.1 Defining a function

We use the keyword `def` to define a function.

General structure:

```python

    # Instructions
```

The code inside the function must be indented.

**Important:** Defining a function does not automatically execute it. We need to call it.

````python
def greet():
  print("Welcome to class")
````

````python
greet()
````

- An **argument** is the actual value supplied when the function is called.

Parameters make functions flexible because we can use the same function with different values.

### 4.4 Understanding `return`

There is an important difference between `print()` and `return`.

- `print()` displays information on the screen.
- `return` sends a result back from a function so that we can use it elsewhere.

Let's compare them.

````python
def calculate_revenue(price, quantity):
    revenue = price * quantity
    return revenue

notebook_revenue = calculate_revenue(1200, 18)

print(notebook_revenue)
print(notebook_revenue + 5000)
````

The function calculates revenue and returns the result.

We store that result in `notebook_revenue`, which means we can use it in another calculation.

Now compare it with a function that only prints the result.

````python
def calculate_and_print_revenue(price, quantity):
    revenue = price * quantity
    print(revenue)

result = calculate_and_print_revenue(1200, 18)

print("Stored result:", result)
````

Notice that `result` is `None`.

The function displayed the revenue, but it did not return the value.

**Remember:** If you need to use the result of a function elsewhere in your programme, use `return`.

### 4.5 Functions with conditional statements

Functions can contain conditional statements.

This means a function can evaluate information and return a decision.

````python
def check_stock(stock):
    if stock <= 10:
        return "Restock"
    else:
        return "Stock is sufficient"

print(check_stock(8))
print(check_stock(25))
````

### 4.6 Combining functions, loops and dictionaries

Let's combine the concepts we have learned so far.

We will:

1. Store product information in a list of dictionaries.
2. Define a function to calculate revenue.
3. Loop through the products.
4. Call the function for each product.
5. Use a conditional statement to check stock.
6. Display the results.

````python
products = [
    {"name": "Notebook", "price": 1200, "quantity_sold": 18, "stock": 8},
    {"name": "Pen", "price": 350, "quantity_sold": 40, "stock": 25},
    {"name": "Calculator", "price": 5500, "quantity_sold": 6, "stock": 4},
    {"name": "School bag", "price": 8000, "quantity_sold": 5, "stock": 12}
]

def calculate_revenue(price, quantity):
    return price * quantity

for product in products:
    revenue = calculate_revenue(
        product["price"],
        product["quantity_sold"]
    )

    if product["stock"] <= 10:
        stock_status = "Restock"
    else:
        stock_status = "Stock is sufficient"

    print("Product:", product["name"])
    print("Revenue: ₦", revenue)
    print("Stock status:", stock_status)
    print("--------------------")
````
