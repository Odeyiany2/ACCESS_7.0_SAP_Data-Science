## 3. Introduction to NumPy

### What is NumPy?

NumPy stands for **Numerical Python**.

It is a Python library designed for working with numerical data. One of its most important features is the NumPy array.

A NumPy array stores values in an organised structure and allows us to perform mathematical operations on many values efficiently.

### Creating a NumPy array

We can create an array from a Python list using `np.array()`.

```python
sales = [1200, 350, 5500, 8000, 2000]

sales_array = np.array(sales)

print(sales_array)
print(type(sales_array))
```

### Understanding the difference between a list and an array

A Python list and a NumPy array can both store multiple values, but they are designed for different kinds of work.

A list is a general-purpose Python collection. A NumPy array is especially useful for numerical operations.

For example, adding a number to a NumPy array adds that number to every element.

```python
sales = [12000, 15000, 16000, 17000, 20000]

print(type(sales))

sales_array = np.array(sales) #convert the list sales to a numpy array

print(type(sales_array))
```

```python
print(sales_array)
```

```python
sales_list = [1200, 350, 5500, 8000, 2000]
sales_array = np.array([1200, 350, 5500, 8000, 2000])

print("List Result:", sales_list * 2)
print("Numpy Array:", sales_array * 2)
```

```python
#print(sales_list + 5000)

print(sales_array + 5000)
```

```python
#print(sales_list + 5000)

new_list = []

for sale in sales_list:
  n_sale = sale + 5000
  new_list.append(n_sale)

print(new_list)
```

```python
sales_list.append(5000)
print(sales_list)
```

**What happened?**

NumPy performed the operation on every value in the array.

- `sales_array + 500` adds 500 to each value.
- `sales_array * 2` multiplies each value by 2.

This is useful when we need to apply the same calculation to many numerical values.

### Creating arrays in different ways

NumPy provides functions that help us create arrays quickly.

```python
# An array containing zeros
zeros_array = np.zeros(5) #create an array that has five zeros
print(zeros_array)
print(zeros_array + 50 * 3)


# An array containing ones
ones_array = np.ones(5)
print(ones_array)

# # An array containing a sequence of numbers
# number_array = np.arange(1, 11)
# print(number_array)
```

```python
#creating a sequence of numbers -> arange
num_array = np.arange(1, 10) # 1 -> 9 1 = starting point, 10 = stopping (stop before it gets to 10)
print(num_array)

num_array_2 = np.arange(0, 21, 2)
print(num_array_2)
```

### Understanding array dimensions and shape

An array can contain values in one or more dimensions.

- A **1-dimensional array** is like a single row of values.
- A **2-dimensional array** is arranged in rows and columns.

The `shape` attribute tells us the size of an array along each dimension.

```python
# One-dimensional array
one_dimensional = np.array([10, 20, 30, 40]) # along one direction - one single row (columns)
print(one_dimensional)
print("Shape:", one_dimensional.shape)
print("Dimensions:", one_dimensional.ndim) # ndim -> tell us the number of dimensions
```

```python
# Two-dimensional array - 2D arrays usually are arranged along anotehr direction  - (rows)
two_dimensional = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [45, 67, 88]
]) #

print(two_dimensional)
print("Shape:", two_dimensional.shape)
print("Dimensions:", two_dimensional.ndim)
```

```python
#three dimensional array
three_dimension = np.array([
    [
        [2,3,4,5],
        [5,6,7,9],

    ],

    [
        [7,8,9,6],
        [5,10,8, 9]
    ]
])

print(three_dimension)
print("Shape:", three_dimension.shape)
print("Dimensions:", three_dimension.ndim)
```

In the second example, the shape is `(2, 3)`. This means the array has 2 rows and 3 columns.

### Accessing values in an array

We use indexing to access individual values. Remember that Python starts counting from index 0.

```python
list_exp = [0, 9, 10, 15]
list_exp[-1]
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[0])
print(numbers[2])
print(numbers[-1])
```

For a two-dimensional array, we provide the row index and column index.

```python
matrix = np.array([
    [10, 20, 30], # 0
    [40, 50, 60] # 1
])

print(matrix[0, 2])  # First row, first column

#print(matrix[1, 2])  # Second row, third column
print(matrix[1, -1])
```

### Basic numerical operations

NumPy can calculate common summary values for an array.

```python
scores = np.array([65, 72, 48, 90, 56])

print("Total:", np.sum(scores))
print("Average:", np.mean(scores))
print("Highest:", np.max(scores))
print("Lowest:", np.min(scores))
```

## 4. Introduction to Pandas

### What is Pandas?

Pandas is a Python library for working with structured data.

It is particularly useful when information is organised in a table, with rows and columns.

For example, a sales dataset may have:
- One row for each order or transaction.
- One column for the order date.
- One column for the product.
- One column for sales.
- One column for profit.

Pandas makes it easier to load, inspect, select, filter and analyse this information.

### The two main Pandas data structures

1. **Series:** A one-dimensional labelled collection of values.
2. **DataFrame:** A two-dimensional table with rows and columns.

Let's look at each one.

### 4.1 Pandas Series

A Series is similar to a single column of data. Each value has an index.

```python
import pandas as pd
```

```python
prices = pd.Series([1200, 350, 5500, 8000])

print(prices)
```

```python
scores = pd.Series(
    [70,85,90],
    index = ["Ada", "John", "Bea"]
)

print(scores)
```

We can also provide labels for the index.

```python
prices = pd.Series(
    [1200, 350, 5500],
    index=["Notebook", "Pen", "Calculator"]
)

print(prices)
print("Price of Pen:", prices["Pen"])
```

### 4.2 Pandas DataFrame

A DataFrame is a table made up of rows and columns.

Let's create a small DataFrame ourselves.

```python
products = {
    "Product": ["Notebook", "Pen", "Calculator", "School bag"],
    "Price": [1200, 350, 5500, 8000],
    "Quantity Sold": [18, 40, 6, 5]
}

product_df = pd.DataFrame(products)

product_df
```

Each column has a name, and each row contains information about one product.

You can think of a DataFrame as a spreadsheet inside Python.

### Series versus DataFrame

- A Series represents one labelled column of values.
- A DataFrame represents a table containing one or more columns.

When we select a single column from a DataFrame, Pandas normally returns a Series.

```python
print(type(product_df["Price"]))
print(type(product_df))
```

```python
product_df["Price"]
```

## 8. Filtering rows

Selecting columns helps us choose which information to view. Filtering helps us choose which records to view based on a condition.

For example, imagine we only want orders where sales are greater than 500.

```python
df_2["Sales"]
```

```python
high_sales = df_2[df_2["Sales"] > 500]

high_sales.shape
```

Let's understand the condition:

`df["Sales"] > 500`

Pandas checks the Sales value in each row and identifies which rows meet the condition.

We then use that condition to filter the DataFrame.

### Filtering with text values

We can also filter using categories.

```python
df_2["Category"]
```

```python
furniture_orders = df_2[df_2["Category"] == "Furniture"]

furniture_orders.head()
```

### Combining conditions

We can combine more than one condition.

Use:
- `&` for AND: both conditions must be true.
- `|` for OR: at least one condition must be true.

Put each condition inside parentheses.

```python
high_sales_and_profit = df_2[
    (df_2["Sales"] > 500) & (df_2["Profit"] > 0)
]

high_sales_and_profit.head()
```

This returns orders where sales are greater than 500 **and** profit is positive.

Now let's use OR.

```python
technology_or_office = df[
    (df["Category"] == "Technology") |
    (df["Category"] == "Office Supplies")
]

technology_or_office.head()
```

This returns records where the category is Technology or Office Supplies.

### Sorting records

We can sort a DataFrame by a column using `sort_values()`.

```python
df.sort_values(by="Sales", ascending=False).head()
```

This displays the five records with the highest sales values.
